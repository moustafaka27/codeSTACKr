```python
world = ['High','Low','Low','High','Low','High','Low','Low']   # High at 0,3,5
n = len(world)
p_hit, p_miss = 0.7, 0.2
p_exact, p_over = 0.8, 0.1
p_under = 1 - p_exact - p_over        # 0.1
U = 2

def sense(p, Z):
    q = [p[i] * (p_hit if world[i] == Z else p_miss) for i in range(n)]
    s = sum(q)
    return [x / s for x in q]          # normalize

def move(p, U):
    q = [0.0] * n
    for i in range(n):
        q[(i + U)     % n] += p_exact * p[i]
        q[(i + U + 1) % n] += p_over  * p[i]
        q[(i + U - 1) % n] += p_under * p[i]
    return q

p = [1/n] * n
for step, Z in [("sense", 'High'), ("move", None), ("sense", 'High'), ("move", None)]:
    p = sense(p, Z) if step == "sense" else move(p, U)
    print(step, [round(x, 4) for x in p])
```
